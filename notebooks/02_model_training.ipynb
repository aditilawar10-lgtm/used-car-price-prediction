```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("../data/processed/cleaned_car_data.csv")
```

```python
print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
print("\nMissing values:")
print(df.isnull().sum().sum())
```

## 2. Separating Features (X) and Target (y)

In supervised machine learning, we separate the input features from the target variable before training the model.

- **Features (X):** The input variables used by the model to make predictions. In this project, they include car age, brand, fuel type, mileage, engine capacity, maximum power, and other relevant car details.
- **Target (y):** The output variable that the model learns to predict. In our project, the target is `selling_price`.

We create X by removing the target and unnecessary columns from the original DataFrame. We create y separately by selecting the `selling_price` column from the original DataFrame.

Keeping X and y separate allows us to train the model using car details and evaluate how accurately it predicts selling prices.

```python
x = df.drop(
    columns=["selling_price", "name", "torque", "year"]
)

y =df["selling_price"]

print("x shape:", x.shape)
print("y shape:", y.shape)
```

## 3. Train-Test Split

The dataset is divided into training and testing sets to evaluate the model's ability to generalize to unseen data.

- **Training set (80%):** Used to train the machine learning model.
- **Testing set (20%):** Used to evaluate the model on unseen observations.

We use `train_test_split()` from Scikit-learn. The parameter `random_state=42` ensures that the same split can be reproduced when the code is run again with the same data and settings.

```python
X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

## 4. Understanding Categorical Features

Categorical features represent groups or labels rather than numerical measurements.

In our Used Car Price Prediction project, the categorical columns are:
- `fuel`
- `seller_type`
- `transmission`
- `owner`
- `brand`
- `mileage_unit`

### Nominal Data

Nominal data consists of categories that have no natural order.

Examples:
- Fuel type: Petrol, Diesel, CNG, LPG
- Transmission: Manual, Automatic
- Brand: Maruti, Hyundai, Toyota, Honda

### Ordinal Data

Ordinal data consists of categories that have a meaningful order or ranking.

For example:
- Low
- Medium
- High

In our dataset, owner categories such as First Owner, Second Owner, and Third Owner have a meaningful sequence. However, we must inspect all actual owner categories before choosing an encoding method.

### Why Do We Need to Handle Categorical Features?

Many machine learning models require numerical inputs. Therefore, categorical values must be represented numerically before training.

Assigning arbitrary numbers to nominal categories can introduce misleading numerical relationships.

For example, encoding Diesel as 1, Petrol as 2, and CNG as 3 may incorrectly suggest an order between fuel types.

### Common Encoding Methods

- **One-Hot Encoding:** Represents categories using separate binary columns. It is commonly used for nominal features.
- **Ordinal Encoding:** Represents ordered categories with numerical values while preserving their defined order.
